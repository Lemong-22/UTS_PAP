```python
import random

print("=== Kuis Matematika ===")
skor = 0
lanjut = 'y'

while lanjut == 'y':
    # 1. Random angka soal
    angka1 = random.randint(1, 20)
    angka2 = random.randint(1, 10)
    
    # 2. Random operator matematika (1: +, 2: -, 3: *)
    # Kita hindari pembagian biar hasilnya nggak desimal ribet
    op_random = random.randint(1, 3)
    
    if op_random == 1:
        operator = '+'
        jawaban_asli = angka1 + angka2
    elif op_random == 2:
        operator = '-'
        jawaban_asli = angka1 - angka2
    else:
        operator = '*'
        jawaban_asli = angka1 * angka2

    # 3. Random pengecoh (range kecil dari jawaban asli)
    pengecoh1 = jawaban_asli + random.randint(1, 5)
    pengecoh2 = jawaban_asli - random.randint(1, 5)
    
    # 4. Random posisi jawaban benar (1: a, 2: b, 3: c)
    posisi_benar = random.randint(1, 3)
    
    print("\nBerapa hasil dari:", angka1, operator, angka2, "?")
    
    # Menampilkan pilihan ganda tanpa list, murni pakai if-else
    if posisi_benar == 1:
        print("a.", jawaban_asli)
        print("b.", pengecoh1)
        print("c.", pengecoh2)
        kunci = 'a'
    elif posisi_benar == 2:
        print("a.", pengecoh1)
        print("b.", jawaban_asli)
        print("c.", pengecoh2)
        kunci = 'b'
    else:
        print("a.", pengecoh1)
        print("b.", pengecoh2)
        print("c.", jawaban_asli)
        kunci = 'c'
        
    # 5. Input dari user
    jawaban_user = input("Pilihanmu (a/b/c): ")
    
    # Cek jawaban
    if jawaban_user == kunci:
        print("Selamat anda benar.")
        skor = skor + 1
    else:
        print("Jawaban anda salah, yang benar adalah ", kunci)
        
    # 6. Tanya mau lanjut atau nggak
    lanjut = input("Lanjutkan soal? (y/n): ")

print("\n=== Kuis Selesai! ===")
print("Skor akhir kamu:", skor)
print("Jancokkk, mantap bro!")
```
